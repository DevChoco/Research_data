```python
from google.colab import files

print("사람 이미지 업로드")
person = files.upload()
print("이미지 업로드")
cloth = files.upload()

person_path = list(person.keys())[0]
cloth_path = list(cloth.keys())[0]
```

```python
%cd leffa
```

```python
from vton_script import LeffaVirtualTryOn

# 인스턴스 생성
vton = LeffaVirtualTryOn(model_type="viton_hd")

# 실행
vton.run(
    person_image_path=person_path,
    garment_image_path=cloth_path,
    output_path="output.jpg",  # 결과 파일 이름
    garment_type="upper_body",
    repaint=False
)
```

```python
from PIL import Image
import matplotlib.pyplot as plt

img = Image.open("output.jpg")
plt.imshow(img)
plt.axis("off")
plt.title("Virtual Try-On Result")
plt.show()
```
